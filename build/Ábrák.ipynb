```python
import numpy as np
import matplotlib.pyplot as plt
import uproot
from scipy.interpolate import interp1d
import xcom
import os

# ----------------------------------------------------------------------
# Betűtípus beállítása

plt.rcParams['figure.figsize'] = (12, 8)
# ----------------------------------------------------------------------

# ======================================================================
# 1. SZIMULÁCIÓS ADATOK BETÖLTÉSE
# ======================================================================

def load_and_sum_histograms(file_list, hist_path="Edep;2", 
                            bin_edges=(0, 3.1, 3000)):
    """
    Betölti a ROOT fájlokat, összeadja a hisztogrammokat.
    
    Returns:
    --------
    hist_sum : numpy array or None
        Összegzett hisztogram (beütésszámok), None ha nem sikerült
    bin_centers : numpy array
        A bin középpontok energiája (MeV)
    """
    n_bins = bin_edges[2]
    bin_min = bin_edges[0]
    bin_max = bin_edges[1]
    
    hist_sum = None
    files_loaded = 0
    
    for file_name in file_list:
        # Ellenőrizzük, hogy a fájl létezik-e
        if not os.path.exists(file_name):
            print(f"FIGYELEM: A fájl nem található: {file_name}")
            continue
            
        try:
            with uproot.open(file_name) as file:
                # Ellenőrizzük, hogy a hisztogramm létezik-e
                if hist_path not in file:
                    print(f"FIGYELEM: A {hist_path} hisztogramm nem található a {file_name} fájlban")
                    print(f"Elérhető kulcsok: {file.keys()}")
                    continue
                    
                hist = file[hist_path]
                
                # Kiolvassuk a hisztogram értékeit
                hist_values = hist.values()
                
                if hist_sum is None:
                    hist_sum = hist_values.copy()
                else:
                    hist_sum += hist_values
                    
                files_loaded += 1
                print(f"Betöltve: {file_name}, beütésszám összeg: {np.sum(hist_values):.0f}")
                
        except Exception as e:
            print(f"Hiba a {file_name} betöltésekor: {e}")
    
    if files_loaded == 0:
        print("ERROR: Egyetlen fájl sem sikerült betölteni!")
        return None, None
    
    # Bin középpontok számítása
    bin_edges_arr = np.linspace(bin_min, bin_max, n_bins + 1)
    bin_centers = (bin_edges_arr[:-1] + bin_edges_arr[1:]) / 2
    
    print(f"\nÖsszesen {files_loaded} fájl betöltve")
    print(f"Teljes beütésszám: {np.sum(hist_sum):.0f}")
    
    return hist_sum, bin_centers

# ----------------------------------------------------------------------
# Fájlok listája (ADD MEG A TELJES ELÉRÉSI ÚTVONALAT!)
# ----------------------------------------------------------------------
# Fontos: itt a pontos elérési utakat kell megadni!
# Például: "C:/Users/.../Radium_empty.root.root" vagy "./Radium_empty.root.root"

data_path = "./"  # Változtasd meg a megfelelő mappára!

empty_files = [
    
    os.path.join(data_path, "Radium_empty_long_1.2mil.root.root")
    
]

alu_15mm_files = [

    os.path.join(data_path, "Radium_1_5cm_long_1.2mil.root.root")
   
]

# ----------------------------------------------------------------------
# Hisztogramok betöltése és összegzése
# ----------------------------------------------------------------------
print("=" * 60)
print("Üres mérés (árnyékolás nélkül)")
print("=" * 60)
empty_hist, bin_centers = load_and_sum_histograms(empty_files)

if empty_hist is None:
    print("Az üres mérés betöltése sikertelen! Kilépés.")
    exit()

print("\n" + "=" * 60)
print("1.5 cm alumínium árnyékolás")
print("=" * 60)
alu_hist, _ = load_and_sum_histograms(alu_15mm_files)

if alu_hist is None:
    print("Az alumínium mérés betöltése sikertelen! Kilépés.")
    exit()

# ----------------------------------------------------------------------
# 2. CSÚCSOK MEGHATÁROZÁSA
# ----------------------------------------------------------------------
def get_peak_areas(hist, bin_centers, energies_MeV, roi_half_width_MeV):
    """
    Kiszámolja a nettó csúcsterületeket a ROI-ban lévő beütések összegzésével.
    """
    if hist is None or bin_centers is None:
        return None, None
        
    peak_areas = []
    peak_areas_err = []
    
    hist_values = hist
    
    for E in energies_MeV:
        # ROI meghatározása
        roi_min = E - roi_half_width_MeV
        roi_max = E + roi_half_width_MeV
        
        # Binek kiválasztása a ROI-ban
        mask = (bin_centers >= roi_min) & (bin_centers <= roi_max)
        roi_indices = np.where(mask)[0]
        
        if len(roi_indices) == 0:
            print(f"Figyelem: nincs bin a {E*1000:.0f} keV csúcs körül!")
            peak_areas.append(0)
            peak_areas_err.append(0)
            continue
        
        # Teljes beütés a ROI-ban
        total_counts = np.sum(hist_values[roi_indices])
        
        # Háttér becslés: a ROI-tól balra és jobbra lévő 5-5 bin átlaga
        left_start = max(0, roi_indices[0] - 20)
        left_end = roi_indices[0]
        right_start = roi_indices[-1] + 1
        right_end = min(len(hist_values), roi_indices[-1] + 11)
        
        left_bkg = np.mean(hist_values[left_start:left_end]) if left_end > left_start else 0
        right_bkg = np.mean(hist_values[right_start:right_end]) if right_end > right_start else 0
        bkg_per_bin = (left_bkg + right_bkg) / 2
        
        # Háttér a ROI-ban
        bkg_counts = bkg_per_bin * len(roi_indices)
        
        # Nettó csúcsterület
        net_area = total_counts -bkg_counts*1.15
        net_area_err = np.sqrt(total_counts + abs(bkg_counts)) if net_area > 0 else np.sqrt(total_counts)
        
        peak_areas.append(net_area)
        peak_areas_err.append(net_area_err)
        
        print(f"{E*1000:.0f} keV: total={total_counts:.0f}, bkg={bkg_counts:.1f}, net={net_area:.0f} ± {net_area_err:.0f}")
    
    return np.array(peak_areas), np.array(peak_areas_err)

# ----------------------------------------------------------------------
# Csúcsterületek kiszámítása
# ----------------------------------------------------------------------
energiak_kev = [186, 295, 352, 609, 1120, 1764, 2204]
energiak_MeV = np.array(energiak_kev) / 1000.0
roi_half_width_keV = 5  # keV
roi_half_width_MeV = roi_half_width_keV / 1000.0

print("\n" + "=" * 60)
print("Üres mérés csúcsterületei")
print("=" * 60)
I_empty, I_empty_err = get_peak_areas(empty_hist, bin_centers, energiak_MeV, roi_half_width_MeV)

if I_empty is None:
    print("Csúcsterület számítás sikertelen!")
    exit()

print("\n" + "=" * 60)
print("1.5 cm Al árnyékolás csúcsterületei")
print("=" * 60)
roi_half_width_keV = 5 # keV

I_alu, I_alu_err = get_peak_areas(alu_hist, bin_centers, energiak_MeV, roi_half_width_MeV)

# ----------------------------------------------------------------------
# Ellenőrizzük, hogy minden csúcs megvan-e
print("\n" + "=" * 60)
print("CSÚCSOK ÖSSZEGZÉSE")
print("=" * 60)
for i, E in enumerate(energiak_kev):
    print(f"{E} keV: Üres: {I_empty[i]:.0f} ± {I_empty_err[i]:.0f}, "
          f"Al: {I_alu[i]:.0f} ± {I_alu_err[i]:.0f}")

# ----------------------------------------------------------------------
# Ha a csúcsterületek túl kicsik, növeljük a ROI-t
if np.max(I_empty) < 100:
    print("\nFIGYELEM: A csúcsterületek nagyon kicsik! Próbáljuk nagyobb ROI-val...")
    roi_half_width_keV = 15
    roi_half_width_MeV = roi_half_width_keV / 1000.0
    print(f"Új ROI félértékszélesség: {roi_half_width_keV} keV")
    
    I_empty, I_empty_err = get_peak_areas(empty_hist, bin_centers, energiak_MeV, roi_half_width_MeV)
    I_alu, I_alu_err = get_peak_areas(alu_hist, bin_centers, energiak_MeV, roi_half_width_MeV)

# ----------------------------------------------------------------------
# A kód folytatása a korábbiak szerint...
# (XCOM adatok, χ² számítás, ábrázolás)
# ----------------------------------------------------------------------
# ----------------------------------------------------------------------

# ======================================================================
# 1. ADATOK ELŐKÉSZÍTÉSE (a szimulációból)
# ======================================================================
# Feltételezve, hogy a korábbi kódban már megvannak:
# I_empty, I_empty_err - üres mérés csúcsterületei
# I_alu, I_alu_err - 1.5 cm Al árnyékolás csúcsterületei
# energiak_kev - a csúcsok energiái

energiak_kev = [186, 295, 352, 609, 1120, 1764, 2204]
n = len(energiak_kev)

# ----------------------------------------------------------------------
# 2. XCOM ADATOK (μ/ρ görbék)
# ----------------------------------------------------------------------
def barn_to_mu_over_rho(sigma_barn, A):
    NA = 6.02214076e23
    return sigma_barn * 1e-24 * NA / A

factory = xcom.MaterialFactory()

# μ/ρ tábla minden Z-re
Mu_S = np.zeros((82, n))  # Z=1..82

for i in range(82):
    data = xcom.calculate_cross_section(i+1)
    energy_eV = np.array(data["energy"], dtype=float)
    energy_MeV = energy_eV / 1e6
    sigma = np.array(data["total"], dtype=float)
    
    A = factory.get_element_mass(i+1)
    mu_rho = barn_to_mu_over_rho(sigma, A)
    
    # Rendezés és interpoláció
    sort_idx = np.argsort(energy_MeV)
    E = energy_MeV[sort_idx]
    MU = mu_rho[sort_idx]
    mask = (E > 0) & (MU > 0)
    E, MU = E[mask], MU[mask]
    
    f_log = interp1d(np.log10(E), np.log10(MU), kind='linear', 
                     bounds_error=False, fill_value=np.nan)
    
    for j, E_keV in enumerate(energiak_kev):
        E_MeV = E_keV / 1000.0
        if E_MeV < E.min() or E_MeV > E.max():
            Mu_S[i, j] = np.nan
        else:
            Mu_S[i, j] = 10**f_log(np.log10(E_MeV))

# ----------------------------------------------------------------------
# 3. FOTOCSÚCS-ARÁNYOS MÓDSZER SZÁMÍTÁSA
# ----------------------------------------------------------------------
rho_x_grid = np.linspace(0.1, 20, 150)
Z_grid = np.arange(1, 83)

# Mátrix a reziduális négyzetek tárolására (csak az 1.5 cm Al-hez)
R2_alu_sim = np.zeros((len(Z_grid), len(rho_x_grid)))

# Logaritmusok előre kiszámolva
ln_I_empty = np.log(I_empty)
ln_I_alu = np.log(I_alu)

# Párok száma
n_pairs = 0
for k1 in range(n):
    for k2 in range(k1+1, n):
        n_pairs += 1

print(f"Energiák száma: {n}, párok száma: {n_pairs}")

# ----------------------------------------------------------------------
# Fő ciklus
# ----------------------------------------------------------------------
print("\nSzámítás folyamatban...")

for zi, Z in enumerate(Z_grid):
    mu_rho_Z = Mu_S[Z-1, :]  # μ/ρ a Z-hez
    
    for ri, rx in enumerate(rho_x_grid):
        res2_sum = 0.0
        
        for k1 in range(n):
            for k2 in range(k1+1, n):
                ln_mert = ln_I_alu[k1] - ln_I_alu[k2]
                ln_ref = ln_I_empty[k1] - ln_I_empty[k2]
                
                delta_mu = mu_rho_Z[k2] - mu_rho_Z[k1]
                theory = delta_mu * rx
                
                # Hiba (Poisson)
                sigma = np.sqrt(
                    (I_empty_err[k1]/I_empty[k1])**2 + (I_empty_err[k2]/I_empty[k2])**2 +
                    (I_alu_err[k1]/I_alu[k1])**2 + (I_alu_err[k2]/I_alu[k2])**2
                )
                
                # Reziduális négyzet
                r2 = ((ln_mert- ln_ref - theory) / sigma)**2
                res2_sum += r2
        
        # Redukált χ²
        R2_alu_sim[zi, ri] = res2_sum / n_pairs

print("Számítás kész.")

# ----------------------------------------------------------------------
# 4. MINIMUM GÖRBE
# ----------------------------------------------------------------------
rx_min_per_Z = np.zeros(len(Z_grid))
for zi, Z in enumerate(Z_grid):
    rx_min_per_Z[zi] = rho_x_grid[np.argmin(R2_alu_sim[zi, :])]

# ----------------------------------------------------------------------
# 5. ÁBRÁZOLÁS
# ----------------------------------------------------------------------
plt.figure(figsize=(12, 8))

# Minimum görbe
plt.plot(rx_min_per_Z, Z_grid, linewidth=1.5, color='green', 
         label='Szimuláció: 1.5 cm Al (Ra-226)')

# Valós érték (alumínium, ρ=2.7 g/cm³, vastagság=1.5 cm)
rho_Al = 2.7
thickness_cm = 0.75
rho_x_real = rho_Al * thickness_cm

plt.scatter([rho_x_real], [13], color='green', s=150, zorder=5, marker='o',
            label=f'Valós: Al {thickness_cm} cm (ρx = {rho_x_real:.2f} g/cm²)')

# Segédvonalak
plt.axvline(rho_x_real, color='green', linestyle='--', alpha=0.4)
plt.axhline(13, color='gray', linestyle='--', alpha=0.4, label='Z = 13 (Al)')

plt.xlabel('Tömegvastagság ρx [g/cm²]', fontsize=12)
plt.ylabel('Rendszám Z', fontsize=12)
plt.legend(fontsize=10)
plt.grid(alpha=0.3)
plt.title('Fotocsúcs-arányos módszer szimulációs adatokon (Ra-226, 1.5 cm Al)', fontsize=12)

plt.savefig('szimulacio_ra226_15mm_Al.pdf', bbox_inches='tight', dpi=300)
plt.savefig('szimulacio_ra226_15mm_Al.png', bbox_inches='tight', dpi=300)
plt.show()

# ----------------------------------------------------------------------
# 6. EREDMÉNYEK KIÍRÁSA
# ----------------------------------------------------------------------
# A minimum görbéből a Z=13-nál lévő ρx érték
Z_Al_idx = 12  # Z=13 (0-index)
rho_x_becsult = rx_min_per_Z[Z_Al_idx]

print("\n" + "=" * 60)
print("EREDMÉNYEK")
print("=" * 60)
print(f"Valós ρx (1.5 cm Al): {rho_x_real:.2f} g/cm²")
print(f"Becsült ρx (Z=13-nál): {rho_x_becsult:.2f} g/cm²")
print(f"Eltérés: {((rho_x_becsult - rho_x_real) / rho_x_real * 100):.1f}%")

# ----------------------------------------------------------------------
# Betűtípus beállítása a folyóirat követelményei szerint

# ----------------------------------------------------------------------

# ======================================================================
# 1. SZIMULÁCIÓS ADATOK (Ra-226)
# ======================================================================
# Feltételezve, hogy a korábbi kódban már megvannak:
# I_empty, I_empty_err - üres mérés csúcsterületei
# I_alu, I_alu_err - 0.75 cm Al árnyékolás csúcsterületei

energiak_kev = [186, 295, 352, 609, 1120, 1764, 2204]
n = len(energiak_kev)

# ----------------------------------------------------------------------
# 2. XCOM ADATOK (μ/ρ görbék)
# ----------------------------------------------------------------------
def barn_to_mu_over_rho(sigma_barn, A):
    NA = 6.02214076e23
    return sigma_barn * 1e-24 * NA / A

factory = xcom.MaterialFactory()

# μ/ρ tábla minden Z-re
Mu_S = np.zeros((82, n))

for i in range(82):
    data = xcom.calculate_cross_section(i+1)
    energy_eV = np.array(data["energy"], dtype=float)
    energy_MeV = energy_eV / 1e6
    sigma = np.array(data["total"], dtype=float)
    
    A = factory.get_element_mass(i+1)
    mu_rho = barn_to_mu_over_rho(sigma, A)
    
    sort_idx = np.argsort(energy_MeV)
    E = energy_MeV[sort_idx]
    MU = mu_rho[sort_idx]
    mask = (E > 0) & (MU > 0)
    E, MU = E[mask], MU[mask]
    
    f_log = interp1d(np.log10(E), np.log10(MU), kind='linear', 
                     bounds_error=False, fill_value=np.nan)
    
    for j, E_keV in enumerate(energiak_kev):
        E_MeV = E_keV / 1000.0
        if E_MeV < E.min() or E_MeV > E.max():
            Mu_S[i, j] = np.nan
        else:
            Mu_S[i, j] = 10**f_log(np.log10(E_MeV))

# ----------------------------------------------------------------------
# 3. FOTOCSÚCS-ARÁNYOS MÓDSZER SZÁMÍTÁSA
# ----------------------------------------------------------------------
rho_x_grid = np.linspace(0.1, 20, 150)
Z_grid = np.arange(1, 83)

# Mátrix a reziduális négyzetek tárolására
R2_alu_sim = np.zeros((len(Z_grid), len(rho_x_grid)))

# Logaritmusok előre kiszámolva
ln_I_empty = np.log(I_empty)
ln_I_alu = np.log(I_alu)

# Párok száma
n_pairs = 0
for k1 in range(n):
    for k2 in range(k1+1, n):
        n_pairs += 1

print(f"Energiák száma: {n}, párok száma: {n_pairs}")

# ----------------------------------------------------------------------
# Fő ciklus (χ² számítás)
# ----------------------------------------------------------------------
print("\nSzámítás folyamatban...")

for zi, Z in enumerate(Z_grid):
    mu_rho_Z = Mu_S[Z-1, :]
    
    for ri, rx in enumerate(rho_x_grid):
        res2_sum = 0.0
        
        for k1 in range(n):
            for k2 in range(k1+1, n):
                ln_mert = ln_I_alu[k1] - ln_I_alu[k2]
                ln_ref = ln_I_empty[k1] - ln_I_empty[k2]
                
                delta_mu = mu_rho_Z[k2] - mu_rho_Z[k1]
                theory = delta_mu * rx
                
                sigma = np.sqrt(
                    (I_empty_err[k1]/I_empty[k1])**2 + (I_empty_err[k2]/I_empty[k2])**2 +
                    (I_alu_err[k1]/I_alu[k1])**2 + (I_alu_err[k2]/I_alu[k2])**2
                )
                
                r2 = ((ln_mert - ln_ref - theory) / sigma)**2
                res2_sum += r2
        
        R2_alu_sim[zi, ri] = res2_sum / n_pairs

print("Számítás kész.")

# ----------------------------------------------------------------------
# 4. HIBASÁVOK SZÁMÍTÁSA (χ² + 1 módszer)
# ----------------------------------------------------------------------
def get_confidence_interval(R2_matrix, rho_x_grid, Z_grid, N_df):
    """
    Visszaadja a minimumok listáját és az alsó-felső hibaintervallumokat minden Z-re.
    χ² + 1 módszer a nem-redukált χ²-re alkalmazva.
    """
    rx_min = np.zeros(len(Z_grid))
    rx_lower = np.zeros(len(Z_grid))
    rx_upper = np.zeros(len(Z_grid))
    chi2_red_min = np.zeros(len(Z_grid))
    
    for zi, Z in enumerate(Z_grid):
        chi2_red_curve = R2_matrix[zi, :]
        
        min_idx = np.argmin(chi2_red_curve)
        rx_min[zi] = rho_x_grid[min_idx]
        chi2_red_min[zi] = chi2_red_curve[min_idx]
        
        threshold = chi2_red_min[zi] + 1.0 / N_df
        
        # Bal oldali keresés
        left_idx = min_idx
        for i in range(min_idx, -1, -1):
            if chi2_red_curve[i] <= threshold:
                left_idx = i
            else:
                break
        
        # Jobb oldali keresés
        right_idx = min_idx
        for i in range(min_idx, len(chi2_red_curve)):
            if chi2_red_curve[i] <= threshold:
                right_idx = i
            else:
                break
        
        rx_lower[zi] = rho_x_grid[left_idx]
        rx_upper[zi] = rho_x_grid[right_idx]
        
        if left_idx == 0 or right_idx == len(chi2_red_curve)-1:
            rx_lower[zi] = np.nan
            rx_upper[zi] = np.nan
    
    return rx_min, rx_lower, rx_upper

# Hibasávok számítása
rx_min, rx_low, rx_up = get_confidence_interval(R2_alu_sim, rho_x_grid, Z_grid, n_pairs)

# ----------------------------------------------------------------------
# 5. ÁBRÁZOLÁS
# ----------------------------------------------------------------------
plt.rcParams['font.sans-serif'] = ['Arial']
plt.rcParams['axes.labelsize'] = 15
plt.rcParams['xtick.labelsize'] = 14
plt.rcParams['ytick.labelsize'] = 14
plt.rcParams['legend.fontsize'] = 16
plt.rcParams['figure.figsize'] = (12, 8)#

# Minimum görbe hibasávval
plt.plot(rx_min, Z_grid, linewidth=1.5, color='green', 
         label='Minimum reziduális görbe (0,75 cm Al)')
plt.fill_betweenx(Z_grid, rx_low, rx_up, color='green', alpha=0.2)

# Valós érték (alumínium, 0.75 cm)
rho_Al = 2.7
thickness_cm = 0.75
rho_x_real = rho_Al * thickness_cm

plt.scatter([rho_x_real], [13], color='green', s=150, zorder=5, marker='o',
            label=r'Valós: Al $0{,}75\,$cm ($\varrho x = 2{,}03$ g/cm²)')

# Segédvonalak
plt.axvline(rho_x_real, color='green', linestyle='--', alpha=0.4)
plt.axhline(13, color='gray', linestyle='--', alpha=0.4, label='Z = 13 (Al)')

plt.xlabel(r'Tömegvastagság $\varrho x$ (g/cm²)',fontsize=22)
plt.ylabel(r'Rendszám $Z$',fontsize=22)
plt.legend(fontsize=20)
plt.grid(alpha=0.3)
#plt.title('Fotocsúcs-arányos módszer szimulációs adatokon (Ra-226, 0,75 cm Al)', fontsize=12)

# ----------------------------------------------------------------------
# Mentés
# ----------------------------------------------------------------------
plt.xlim(0,3)
plt.savefig('abra/szimulacio_ra226_075cm_Al_hibasav.pdf', bbox_inches='tight', dpi=300)
plt.savefig('abra/szimulacio_ra226_075cm_Al_hibasav.png', bbox_inches='tight', dpi=300)
plt.show()

# ----------------------------------------------------------------------
# 6. EREDMÉNYEK KIÍRÁSA
# ----------------------------------------------------------------------
Z_Al_idx = 12  # Z=13 (0-index)
rho_x_becsult = rx_min[Z_Al_idx]
rho_x_becsult_low = rx_low[Z_Al_idx]
rho_x_becsult_high = rx_up[Z_Al_idx]
```
